## Business case:- On the basis of given parameters, our target is to find which crop is suitable for the given agricultural conditions. So we are solving a classification task.

```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
```

```python
data = pd.read_csv("Crop_recommendation.csv")
print(data.shape)
data.head()
```

```python
X = data.iloc[:,:-1]
y = data.label
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
sclaed_x=sc.fit_transform(X)
X=pd.DataFrame(sclaed_x,columns=X.columns)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=10)
```

```python
from sklearn.neural_network import MLPClassifier
model=MLPClassifier(hidden_layer_sizes=(50,3),learning_rate_init=0.1,max_iter=100,random_state=2)
model.fit(X_train,y_train)
y_predict=model.predict(X_test)
y_train_predict=model.predict(X_train)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
print("Train accuracy :",accuracy_score(y_train,y_train_predict))
print("Test accuracy :",accuracy_score(y_test,y_predict))
```
